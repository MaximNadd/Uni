```python
import pandas as pd
import numpy as np

df = pd.read_excel('dataset.xlsx', engine='openpyxl')
df = df.iloc[:, :-1]
df.head(5)
```

```python
column_list = df.columns.tolist()
column_list = list(df.columns)
column_list = column_list[2:]
column_list
```

```python
df['общие_расходы'] = df[column_list].sum(axis=1)
df.head(500)
```

```python
conditions = [
    df['общие_расходы'] < 150000,
    (df['общие_расходы'] >= 150000) & (df['общие_расходы'] < 180000),
    df['общие_расходы'] >= 180000
]
df['категория_дохода'] = np.select(conditions, ['Низкий', 'Средний', 'Высокий'])
df.head(10)
```

```python
df['детский_сад'] = np.where(df['Кол-во Детей'] > 0, 'да', 'нет')
df.head(10)
```
