```python
import pandas as pd
import numpy as np

df = pd.read_excel('dataset.xlsx', engine='openpyxl')
df = df.iloc[:, :-1]
df.head(5)
```

```python
column_list = df.columns.tolist()
column_list = list(df.columns)
column_list = column_list[2:]
column_list
```

```python
df['общие_расходы'] = df[column_list].sum(axis=1)

conditions = [
    df['общие_расходы'] < 3000000,
    (df['общие_расходы'] >= 3000000) & (df['общие_расходы'] < 8000000),
    df['общие_расходы'] >= 8000000
]
df['категория_дохода'] = np.select(conditions, ['Low Income', 'Middle Income', 'High Income'])
df.head(500)
```
